```python
from config import output_file, margin_file
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
```

```python
df1 = pd.read_csv(margin_file("margins1_cit.csv"))
df2 = pd.read_csv(margin_file("margins2_cit.csv"))
df3 = pd.read_csv(margin_file("margins3_cit.csv"))
df4 = pd.read_csv(margin_file("margins4_cit.csv"))
```

```python
df1_ = pd.read_csv(margin_file("margins1_cd.csv"))
df2_ = pd.read_csv(margin_file("margins2_cd.csv"))
df3_ = pd.read_csv(margin_file("margins3_cd.csv"))
df4_ = pd.read_csv(margin_file("margins4_cd.csv"))
```

```python
dflst = [df1 , df2 , df3 , df4]
```

```python
dflst_ = [df1_ , df2_ , df3_ , df4_]
```

```python
fig = plt.figure(figsize=(12 , 10))
gs = gridspec.GridSpec(2, 4, figure=fig, wspace=.25, hspace=.25)

for i in range(4):
    ax = fig.add_subplot(gs[i], projection='3d')
    df = dflst[i]
    trisurf = ax.plot_trisurf(df['b_cum_k'], df['age_diff_abs'], df['estimate'], cmap=plt.cm.Spectral_r)
    
    ax.set_xlabel('$k (t)$')
    ax.set_ylabel('Age diff')
    ax.set_xlim([df['b_cum_k'].min(), df['b_cum_k'].max()])
    ax.set_ylim([df['age_diff_abs'].min(), df['age_diff_abs'].max()])
    ax.set_zlim([df['estimate'].min(), df['estimate'].max()])

for i in range(4):
    ax = fig.add_subplot(gs[4+i], projection='3d')
    df = dflst_[i]
    trisurf = ax.plot_trisurf(df['b_cum_k'], df['age_diff_abs'], df['estimate'], cmap=plt.cm.Spectral_r)
    
    ax.set_xlabel('$k (t)$')
    ax.set_ylabel('Age diff')
    ax.set_xlim([df['b_cum_k'].min(), df['b_cum_k'].max()])
    ax.set_ylim([df['age_diff_abs'].min(), df['age_diff_abs'].max()])
    ax.set_zlim([df['estimate'].min(), df['estimate'].max()])

plt.savefig(output_file("margins.svg"),bbox_inches = 'tight')
```
